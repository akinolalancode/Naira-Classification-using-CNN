```python
import gradio as gr
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import img_to_array
import numpy as np
from gtts import gTTS
import cv2

# Load class labels
class_labels = ["10", "100", "1000", "20", "200", "5", "50", "500"]

def preprocess_image(image):
    image = cv2.resize(np.array(image), (224, 224))
    image = img_to_array(image)
    image = np.expand_dims(image, axis=0)
    return image / 255.0

def predict(model_name, image):
    model_path = f"saved_models/{model_name}_naira_model.h5"
    model = load_model(model_path)
    preprocessed = preprocess_image(image)
    pred = model.predict(preprocessed)
    label_index = np.argmax(pred)
    label = class_labels[label_index]
    response = f"Predicted denomination: {label} NAIRA"
    
    tts = gTTS(response)
    tts.save("output.mp3")
    return response, "output.mp3"

# Gradio Interface
interface = gr.Interface(
    fn=predict,
    inputs=[
        gr.Dropdown(choices=["Basic_CNN", "VGG19", "ResNet50", "InceptionV3"], label="Choose Model"),
        gr.Image(label="Upload Naira Note Image")
    ],
    outputs=[
        gr.Textbox(label="Prediction"),
        gr.Audio(label="Audio Output", type="filepath")
    ],
    title="naira classification developed by AKINOLA, Lanre Olusegun",
    theme="soft"
)

interface.launch(share=True)
```
